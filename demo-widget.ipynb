```python
%env ANYWIDGET_HMR=1
import anywidget
import traitlets


class DemoWidget(anywidget.AnyWidget):

    def __init__(self, colour='red', **kwargs):
        super().__init__(**kwargs)
        self.colour = colour

    _esm = "src/script.js";
    # _css = "src/gui/styles.css"

    colour = traitlets.Unicode().tag(sync=True)
    x_value = traitlets.Float().tag(sync=True)
    y_value = traitlets.Float().tag(sync=True)
```

```python
red_widget = DemoWidget(colour='red')
red_widget
```

```python
blue_widget = DemoWidget(colour='blue')
blue_widget
```

```python
# print coordinates of last dot
print(red_widget.x_value, red_widget.y_value)
print(blue_widget.x_value, blue_widget.y_value)
```
